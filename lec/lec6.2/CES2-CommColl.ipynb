```python
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px 

```

```python
#cd EnviroScreen4
CES4 = pd.read_excel("calenviroscreen40resultsdatadictionary_F_2021.xlsx", sheet_name='CES4.0FINAL_results')
```

```python
CES4
```

```python
CollegeCodes= pd.read_csv("colleges.csv")
CollegeCodes
```

```python
#merge CollegeCodes_Public with CES4
CES4_Public = pd.merge(CES4, CollegeCodes, how='inner', left_on='ZIP', right_on='Zip')
```

```python
# Find the most polluted zip codes and show the college there
CES4_Public.sort_values(by='CES 4.0 Score', ascending=False).head(10)
```

```python
# Find the least polluted zip codes and show the college there
CES4_Public.sort_values(by='CES 4.0 Score', ascending=True).head(10)
```

```python
# what is the score at El Camino College
CES4_Public[CES4_Public['College'] == 'EL CAMINO COLLEGE']
```

```python
# histogram of the CES4 scores for the colleges
plt.hist(CES4_Public['CES 4.0 Score'], bins=20)
plt.xlabel('CES 4.0 Score')
plt.ylabel('Number of Colleges')
plt.title('Distribution of CES 4.0 Scores for Colleges')
plt.show()
```

```python
#descibe the distribution of CES4 scores for the colleges
CES4_Public['CES 4.0 Score'].describe()
```

```python
# What is the range of CES4 scores for the colleges?
CES4_Public['CES 4.0 Score'].min(), CES4_Public['CES 4.0 Score'].max()
```

```python
# plotly map of the CES4 scores for the colleges
fig = px.scatter_map(CES4_Public, lat="Latitude", lon="Longitude", hover_name="College", hover_data=["CES 4.0 Score"], color="CES 4.0 Score", color_continuous_scale=px.colors.sequential.Viridis, zoom=5, height=600)
fig.update_layout(mapbox_style="open-street-map")
fig.update_layout(margin={"r":0,"t":0,"l":0,"b":0})
fig.show()
```

```python
# Map of CES scores for colleges scoring above 50
CES4_Public_High = CES4_Public[CES4_Public["CES 4.0 Score"] > 50]
fig = px.scatter_map(
    CES4_Public_High,
    lat="Latitude",
    lon="Longitude",
    hover_name="College",
    hover_data=["CES 4.0 Score"],
    color="CES 4.0 Score",
    color_continuous_scale=px.colors.sequential.Viridis,
    zoom=5,
    height=600,
)
fig.update_layout(mapbox_style="open-street-map")
fig.update_layout(margin={"r":0,"t":0,"l":0,"b":0})
fig.show()
```
